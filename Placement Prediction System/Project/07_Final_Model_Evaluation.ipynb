## 11. Verify Train/Test Feature Consistency

The feature columns of the final training and testing datasets are compared to ensure that they are identical and appear in the same order.

The final Decision Tree model will be trained using the 12 selected features from the training dataset and will later receive the testing dataset as input. Therefore, the training and testing datasets must have exactly the same feature structure.

This verification checks:

* The number of features in both datasets.
* Whether the feature names are identical.
* Whether the feature columns appear in the same order.
* Whether any feature is present in one dataset but missing from the other.

A successful consistency check confirms that the testing data has the correct structure for final model evaluation.


````python
print("Train/Test Feature Consistency Check")
print("=" * 60)

train_features = X_train.columns.tolist()
test_features = X_test.columns.tolist()

print("Number of training features :", len(train_features))
print("Number of testing features  :", len(test_features))

print("\nFeature names identical:", train_features == test_features)

missing_in_test = set(train_features) - set(test_features)
missing_in_train = set(test_features) - set(train_features)

print("\nFeatures missing in testing dataset:", missing_in_test)
print("Features missing in training dataset:", missing_in_train)

feature_consistency = (
    train_features == test_features
    and len(missing_in_test) == 0
    and len(missing_in_train) == 0
)

print("\nFeature consistency:", feature_consistency)

if feature_consistency:
    print("Training and testing feature structures are consistent.")
else:
    print("Training and testing feature structures are inconsistent.")
````

## 22. Calculate Overall Performance Metrics

The final Decision Tree predictions are evaluated using five overall classification metrics.

The metrics used are:

1. **Accuracy** — measures the proportion of all correctly classified samples.
2. **Weighted Precision** — measures the precision of predictions while accounting for the number of samples in each class.
3. **Weighted Recall** — measures the recall across all classes while weighting each class according to its support.
4. **Weighted F1-score** — provides a weighted balance between precision and recall.
5. **Macro F1-score** — calculates the F1-score independently for each class and then gives equal importance to every class.

Macro F1-score is particularly important for this dataset because the target classes are highly imbalanced. The minority class must therefore be considered separately rather than allowing the majority class to dominate the overall evaluation.

The actual target values from `y_test` are compared with the predictions generated by the final Decision Tree model.


````python
print("Final Model Overall Performance Evaluation")
print("=" * 70)

y_true = y_test[target_column]

final_accuracy = accuracy_score(
    y_true,
    y_pred
)

final_weighted_precision = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

final_weighted_recall = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

final_weighted_f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

final_macro_f1 = f1_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

print(f"Accuracy           : {final_accuracy:.6f}")
print(f"Weighted Precision : {final_weighted_precision:.6f}")
print(f"Weighted Recall    : {final_weighted_recall:.6f}")
print(f"Weighted F1-score  : {final_weighted_f1:.6f}")
print(f"Macro F1-score     : {final_macro_f1:.6f}")
````

## 28. Error Analysis

Error analysis examines the predictions that do not match the actual target classes.

For the final Decision Tree model, the confusion matrix is used to identify the number of correct predictions and misclassifications for each class.

The analysis focuses on:

* Total number of correct predictions.
* Total number of incorrect predictions.
* Number of errors for each actual class.
* Number of correctly classified samples for each actual class.
* The classes involved in misclassification.

This analysis is particularly important for the CIC-IDS2017 dataset because the target classes are imbalanced. A model can achieve very high overall accuracy while still making a comparatively larger proportion of errors on a minority class.

The error analysis therefore complements the overall accuracy, weighted metrics, and Macro F1-score by examining the classification behavior of individual classes.


````python
print("Final Model Error Analysis")
print("=" * 70)

# Total correct and incorrect predictions
total_samples = len(y_true)

correct_predictions = np.trace(final_confusion_matrix)
incorrect_predictions = total_samples - correct_predictions

print(f"Total test samples       : {total_samples:,}")
print(f"Correct predictions      : {correct_predictions:,}")
print(f"Incorrect predictions    : {incorrect_predictions:,}")

print("\nOverall prediction analysis:")
print(f"Correct prediction rate  : {correct_predictions / total_samples:.6f}")
print(f"Error rate               : {incorrect_predictions / total_samples:.6f}")

# Class-wise error analysis
class_labels = [0, 1, 2, 3]

error_analysis = []

for index, class_label in enumerate(class_labels):
    actual_count = final_confusion_matrix[index].sum()
    correct_count = final_confusion_matrix[index, index]
    error_count = actual_count - correct_count
    
    error_rate = error_count / actual_count if actual_count > 0 else 0
    
    error_analysis.append({
        "Class": class_label,
        "Actual Samples": actual_count,
        "Correct Predictions": correct_count,
        "Incorrect Predictions": error_count,
        "Error Rate": error_rate
    })

error_analysis_df = pd.DataFrame(error_analysis)

print("\nClass-Wise Error Analysis:")
print(error_analysis_df.to_string(index=False))

print("\nError analysis completed successfully.")
````

````python
print("FINAL MODEL CHARACTERISTICS AND SUMMARY")
print("=" * 70)

# Recalculate final evaluation metrics
accuracy = accuracy_score(y_true, y_pred)

weighted_precision = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

weighted_recall = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

weighted_f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

macro_f1 = f1_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

print("\nModel Information")
print("-" * 70)
print("Model type              : Decision Tree Classifier")
print("Class weight            : balanced")
print("Random state            : 42")
print("Number of features      :", X_train.shape[1])

print("\nDataset Information")
print("-" * 70)
print(f"Training samples        : {X_train.shape[0]:,}")
print(f"Testing samples         : {X_test.shape[0]:,}")

print("\nComputational Performance")
print("-" * 70)
print(f"Training time           : {final_training_time:.2f} seconds")
print(f"Prediction time         : {final_prediction_time:.2f} seconds")

print("\nFinal Evaluation Metrics")
print("-" * 70)
print(f"Accuracy                : {accuracy:.6f}")
print(f"Weighted Precision      : {weighted_precision:.6f}")
print(f"Weighted Recall         : {weighted_recall:.6f}")
print(f"Weighted F1-score       : {weighted_f1:.6f}")
print(f"Macro F1-score          : {macro_f1:.6f}")

print("\nClass-Wise F1 Scores")
print("-" * 70)

for class_name, row in class_wise_results.iterrows():
    print(f"{class_name:<10}: {row['f1-score']:.6f}")

print("\nFinal model summary completed successfully.")
````
